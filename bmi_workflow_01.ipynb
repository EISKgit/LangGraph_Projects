```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict    
```

```python
#define the State
class BMIState(TypedDict):
    weight: float
    height: float
    bmi: float
    label: str
    
```

```python
def calculate_bmi(state=BMIState) -> BMIState:
    state['bmi'] = state['weight'] / (state['height'] ** 2)
    return state
```

```python
def label_bmi(state=BMIState) -> BMIState:
    bmi = state['bmi']
    if bmi < 18.5:
        state['label'] = 'Underweight'
    elif 18.5 <= bmi < 24.9:
        state['label'] = 'Normal weight'
    elif 25 <= bmi < 29.9:
        state['label'] = 'Overweight'
    else:
        state['label'] = 'Obesity'
    return state
```

```python
# define the graph
bmi_graph = StateGraph(BMIState)

# add nodes to the graph
bmi_graph.add_node("calculate_bmi",calculate_bmi)
bmi_graph.add_node("label_bmi",label_bmi)

# add edges to the graph
bmi_graph.add_edge(START, "calculate_bmi")
bmi_graph.add_edge("calculate_bmi", "label_bmi")
bmi_graph.add_edge("label_bmi", END)
# compile the graph
workflow = bmi_graph.compile()
```

```python
# execute the graph
initial_state = {'weight': 75, 'height': 1.75}
final_state = workflow.invoke(initial_state)
print(final_state)
```
